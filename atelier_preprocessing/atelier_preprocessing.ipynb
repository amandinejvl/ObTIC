## 1. Pourquoi prétraiter des données textuelles ?

### **(Très) brève introduction au TAL/NLP**

Le **NLP** (_Natural Language Processing_), ou en français **TAL** (_Traitement automatique des langues_) ou parfois **TALN** (_Traitement automatique du Langage Naturel_), est la discipline qui consiste à faire comprendre, analyser, manipuler, et générer du langage humain par des ordinateurs.

<img src="https://content-media-cdn.codefinity.com/courses/c68c1f2e-2c90-4d5d-8db9-1e97ca89d15e/section_1/chapter_1/nlp_illustration.png?w=1920&q=75" alt="POS-tagging" width="600">


Dans de nombreux domaines des Sciences Humaines et Sociales, on dispose souvent de **corpus textuels** dont on aimerait pouvoir **automatiser le traitement**.

Or, ces corpus bruts ne sont pas toujours prêts pour l'analyse ou le traitement souhaité.

Il faut donc dans ce cas recourir à une étape de **prétraitement**.

Ce prétraitement répond à plusieurs objectifs ➡️

#### **✨ Normalisation ✨**

Convertir un texte brut en une forme standard qui simplifie son traitement et son analyse par des **systèmes automatisés**.

- Réduire le bruit et les variations (ex : orthographe, ponctuation, casse, etc.)
- Homogénéiser des textes pour les rendre comparables

Nécessaire pour de nombreuses tâches de **NLP** :

- Extraction d'informations
- Comparaison de textes
- Classification de textes
- etc.

#### **Exemples de normalisations**

- **Suppression des éléments non significatifs** : ponctuation, _stopwords_
- **Homogénéisation** : mise en minuscule, lemmatisation
- **Segmentation** : découper le texte en unités pertinentes selon la tâche (lignes, phrases, mots, etc.)

#### **✨ Textométrie ✨**


La **textométrie** est la discipline qui mesure et compare statistiquement des textes. Elle peut avoir pour objets d'étude les occurrences des mots, leurs formes, les structures morphosyntaxiques, etc.

L'objectif est d'objectiver des observations par des mesures **quantifiables** et **statistiques**.


#### **Exemples de projets impliquant de la textométrie**

- Analyse de l'évolution du vocabulaire lié à un sujet dans la presse
- Analyse des réécritures d'une oeuvre
- Analyse stylistique
- Evolution des concepts par étude de co-occurrences
- etc.

### **2.1. Tâches de prétraitement de données textuelles brutes**

Les tâches de **prétraitement** et de **préparation** de données textuelles brutes sont nombreuses. Le traitement à appliquer au corpus **dépend des objectifs** de recherche et de la **méthodologie** envisagée. Il peut s'agir de nettoyage, de segmentation, ou encore d'étiquetage des données.

Nous listerons ici les principales tâches de prétraitement et de préparation de données textuelles.

#### 1️⃣ **Nettoyage**

#### **Mise en minuscules**

La **mise en minuscules** d'un texte consiste à convertir toutes les lettres d'un texte en minuscules. Cette méthode élimine les distinctions de **casse**, par exemple entre 'Paris', 'paris' et 'PARIS', pour les traiter comme un seul et même mot. Cela **uniformise** le texte, simplifie les **comparaisons** et est particulièrement utile dans les tâches de recherche ou d'analyse où la différence entre majuscules et minuscules n'a pas d'impact sur le sens. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

<img src="https://uxwing.com/wp-content/themes/uxwing/download/editing-user-action/uppercase-to-lowercase-icon.png" alt="maj_to_min" width="100">

Par exemple, dans le comptage des occurrences d'un mot dans un texte, le fait d'appliquer ou non ce prétraitement aura une influence sur les résultats.


**Exemple :**

> _Un jour un coq détourna Une perle, qu'il donna Au beau premier lapidaire._

**"Un": 1** | "jour": 1 | **"un": 1** | "coq": 1 | "détourna": 1 | etc.

⬇️

> _un jour un coq détourna une perle, qu'il donna au beau premier lapidaire._

#### **Suppression de la ponctuation**

Une autre tâche élémentaire du nettoyage de texte est la **suppression de la ponctuation**. Là encore, cette étape n'est pas nécessaire et dépend de notre objectif.

Même si la ponctuation joue un rôle essentiel dans la structuration et la clarté des phrases, dans certaines tâches de NLP, elle peut être supprimée pour **simplifier** les données et réduire les variations inutiles.

<img src="https://filestore.medicineslearningportal.org/images/istock%20punctuation.png" alt="punct" width="200">

Cette étape est particulièrement utile dans des applications telles que l'**analyse statistique**, la classification ou la création de modèles d'apprentissage automatique, car elle permet de se concentrer sur les éléments de contenu **significatifs**. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

**Exemple :**

> _Ah ! non ! c’est un peu court, jeune homme !
On pouvait dire… Oh ! Dieu ! … bien des choses en somme…_

⬇️

> Ah non c’est un peu court jeune homme
On pouvait dire Oh Dieu bien des choses en somme

#### **Suppression des mots-vides (_stopwords_)**

Les **mots-vides (ou _stopwords_)** sont des mots qui ne sont pas porteurs d'une grande information sémantique. Ce sont généralement les mots grammticaux comme _de_, _qui_, _le_, etc.

<img src="https://repository-images.githubusercontent.com/70948862/f8d8f380-beb0-11e9-8f03-050093c4220c" alt="Stopwords" width="400">

La **suppression des mots-vides (_stopwords_)** est une technique couramment utilisée pour limiter le **bruit** dans les tâches d'analyse textuelle. Ces mots, bien qu'indispensables à la construction grammaticale, apportent peu de valeur sémantique et risquent d'**alourdir** les analyses. En les éliminant, on peut non seulement simplifier les représentations textuelles, mais aussi optimiser la taille des structures de données. Cela permet de focaliser les efforts de traitement sur les termes porteurs de sens et d'améliorer la pertinence des résultats. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

**Exemple :**

> _Car vouloir faire ce qui passe nos forces, c'est de la démence._

⬇️

> _vouloir faire passe forces démence_

#### **Lemmatisation**

La **lemmatisation** est une composante importante de la **normalisation**, qui vise à réduire les mots à leur **forme de base ou '_lemme_'**. Cela permet de regrouper différentes formes d'un même mot sous une seule représentation. Par exemple, "_sang_", "_sung_" et "_sings_" en anglais sont toutes des variantes du verbe "_sing_" et seront ramenées à cette forme canonique.

<img src="https://miro.medium.com/v2/resize:fit:640/format:webp/1*A5YJhglv0NXW9kNJBA0-8w.jpeg" alt="Lemmatisation" width="400">

La lemmatisation est particulièrement utile pour **réduire la complexité lexicale**, améliorer la généralisation et obtenir des analyses plus précises dans des tâches comme la recherche ou la classification. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

**Exemple :**

> _Ma mère-grand, que vous avez de grands bras ! — C’est pour mieux t’embrasser, ma fille !_

⬇️

> _mon mère grand, que vous avoir de grand bras ! — c’être pour mieux t’ embrasser, mon fille !_

#### **Segmentation en phrases**

La **segmentation en phrases** consiste à diviser un texte en unités syntaxiques appelées phrases, en utilisant des indices comme la ponctuation. Cette méthode est cruciale pour de nombreuses applications en NLP, telles que la traduction automatique, la classification ou l'analyse de sentiments. Voici quelques aspects essentiels de ce processus.

(**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

#### **Segmentation en lignes**

La **segmentation en lignes** consiste à diviser un texte en unités sur la base des sauts de ligne. Contrairement à la segmentation par phrases, elle s'appuie sur la structure visuelle ou formatée du texte, souvent marquée par des retours à la ligne.

Cette technique est couramment utilisée dans des contextes où les phrases ne sont pas naturellement délimitées par la ponctuation, comme dans les listes, les poèmes ou certains documents formalisés (CV, rapport, formulaire, etc.). (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

**Exemple :**

Lignes :

> _Ô fins d’automne, hivers, printemps trempés de boue,_

> _Endormeuses saisons ! je vous aime et vous loue_

> _D’envelopper ainsi mon cœur et mon cerveau_

> _D’un linceul vaporeux et d’un brumeux tombeau._

Phrases :

> _Ô fins d’automne, hivers, printemps trempés de boue, Endormeuses saisons !_

> _je vous aime et vous loue D’envelopper ainsi mon cœur et mon cerveau D’un linceul vaporeux et d’un brumeux tombeau._

#### **Tokenisation**

La **tokenisation** consiste à diviser un texte en unités appelées "_tokens_", généralement des mots.

<img src="https://miro.medium.com/v2/resize:fit:828/format:webp/1*EmXiN_xslbFyBmvxCT_vpw.png" alt="Tokenisation" width="500">

Bien que les mots en français ou en anglais soient généralement séparés par des espaces, cela ne suffit pas toujours. Par exemple, des expressions comme "_pomme de terre_", "_New York_" ou "_rock n' roll_" peuvent être traitées comme des entités uniques malgré les espaces qu'elles contiennent. À l'inverse, des contractions comme "_I'm_" doivent être divisées en deux tokens : "_I_" et "_am_". La tokenisation est cruciale car la plupart des analyses linguistiques nécessitent des données segmentées de manière cohérente. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_preprocessing)_)

**Exemple :**

> _Il y avait une table servie sous un arbre devant la maison, et le Lièvre y prenait le thé avec le Chapelier._

⬇️

> _Il_ | _y_ | _avait_ | _une_ | _table_ | _servie_ | _sous_ | _un_ | _arbre_ | _devant_ | _la_ | _maison_ | _,_ | _et_ | _le_ | _Lièvre_ | _y_ | _prenait_ | _le_ | _thé_ | _avec_ | _le_ | _Chapelier_ | _._

#### 3️⃣ **Etiquetage**

#### **Etiquetage morphosyntaxique (_POS-tagging_)**

L'**étiquetage morphosyntaxique**, également appelé _Part-of-Speech tagging_ , est une méthode de traitement automatique du langage naturel (NLP) qui consiste à attribuer une catégorie grammaticale à chaque mot d'un texte. Cette tâche est fondamentale pour analyser et structurer le contenu linguistique de manière automatisée. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_annotation)_)

<img src="https://miro.medium.com/v2/resize:fit:828/format:webp/1*qZELwIpKeEQ-j3EnRF-CrQ.jpeg" alt="POS-tagging" width="450">


Cet étiquetage fournit une analyse linguistique détaillée qui peut être utilisée dans diverses applications :

- **Comparaison de traductions et éditions :** Identifier les différences entre deux versions d'un même texte, notamment au niveau grammatical.
- **Filtrage de termes :** Extraire les verbes, noms propres ou toute autre catégorie pour des analyses spécifiques.
- **Prétraitement pour des tâches complexes :** Les modèles avancés comme ceux de traduction automatique ou d'analyse sémantique dépendent fortement de cette catégorisation.

(**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_annotation)_)

**Exemple :**

> _Le château de Dracula se dessinait à présent, net, contre le ciel rouge et chaque pierre de ses créneaux brisés prenait un relief effrayant, inoubliable._

⬇️

> _Le_ : DET | _château_ : NOUN | _de_ : ADP | _Dracula_ : PROPN | _se_ : PRON | _dessinait_ : VERB | _à_ : ADP | _présent_ : NOUN | _,_ : PUNCT | etc.

#### **Reconnaissance d'entités nommées (NER)**

La **reconnaissance des entités nommées** (REN), ou _Named Entity Recognition_ (NER) en anglais, est une tâche clé du traitement automatique du langage naturel (NLP). Elle consiste à identifier dans un texte des portions spécifiques qui correspondent à des entités et à leur attribuer un type particulier. Une entité nommée, de manière générale, désigne tout élément qui peut être référencé par un **nom propre**, comme une personne, un lieu ou une organisation. Cependant, le concept d'entité nommée peut être élargi pour inclure des dates, des expressions temporelles, des prix ou d'autres formes d'expressions numériques. (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_annotation)_)

<img src="https://miro.medium.com/v2/resize:fit:828/format:webp/1*25pE_3FRfkkABhcFR2UAcA.png" alt="Stopwords" width="600">

La REN est utilisée dans de nombreux domaines et applications :

- **Extraction d'information :** Identifier automatiquement des informations clés dans de vastes volumes de données, comme les noms des intervenants dans un texte ou les lieux d'un événement.
- **Systèmes de recherche d'information :** Améliorer la recherche documentaire en focalisant sur les entités nommées plutôt que sur des mots-clés généraux.
- **Analyse de sentiments :** Examiner les opinions ou sentiments associés à des entités spécifiques.
- **Traduction automatique :** Reconnaître et traduire correctement les noms propres ou entités spécifiques d'un texte source vers une langue cible.
- **Reconnaissance vocale :** Mieux interpréter les noms propres ou termes uniques dans les commandes vocales.

(**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_annotation)_)

**Exemple :**

> _Le juge Wargrave, retraité depuis peu, tirait sur son cigare en parcourant avec intérêt les pages politiques du Times._

⬇️

> _Wargrave_ : PERSONNE

> _Times_ : ORGANISATION

#### **Syntaxe en dépendance (_Dependency parsing_)**

L'**analyse des dépendances syntaxiques** décrit la structure grammaticale d'une phrase en mettant en relation les mots sous la forme de relations binaires entre un nœud central et un dépendant.

La phrase est représentée comme un arbre de dépendance possédant un seul noeud racine (généralement le verbe). Tous les autres mots ont exactement un lien sortant vers leur tête grammaticale. Les relations décrivent des concepts grammaticaux comme le sujet, l'objet ou les modificateurs (ex. : adjectifs, adverbes). (**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_analyses)_)

<img src="https://spacy.io/images/displacy.svg" alt="Dependency parsing" width="600">


**Applications principales :**

- **Analyse syntaxique avancée** : Étudier la structure des phrases pour en extraire des informations grammaticales.

- **Traduction automatique** : Maintenir la cohérence grammaticale lors de la traduction.

- **Extraction d'information** : Identifier des relations spécifiques, comme 'qui fait quoi'.

(**Source :** _[Pandore Toolbox](https://obtic-gpu1.mesu.sorbonne-universite.fr/pandore/documentation_analyses)_)

##### **⚙️ Automatisation**

##### **🚀 Performances**

- Les interfaces graphiques atteignent rapidement leurs **limites** face à de très **gros corpus** (mémoire, temps de calcul, blocages).

- Python permet un traitement généralement plus **optimisé**, qui permet de traiter plus de documents, en un temps réduit.

> **Exemple :** Analyser l'ensemble des oeuvres d'un auteur, des documents individuellement longs, et nombreux, qui peut vite faire planter une interface ou un serveur web.

##### **🔄 Reproductibilité**

- Un script **documente** de lui-même chaque étape et chaque paramètre du traitement, assurant sa reproductibilité.

- Contrairement aux clics dans une interface, le code peut être versionné, commenté, réexécuté **à l'identique**.

> **Exemple :** Un script Python doit permettre d'obtenir les mêmes résultats sur les mêmes données, peut importe qui l'exécute.

##### **3. Les chaînes de caractères (`str`)**

Ce sont les objets qu'on va manipuler en permanence en NLP.

````python
phrase = "  Le Traitement Automatique du Langage, c'est SUPER ! "

print(phrase.lower())                           # tout en minuscules
print(phrase.strip())                           # enlève les espaces au début et à la fin
print(phrase.strip().lower().replace("!", ""))  # concaténation des méthodes
print(phrase.split())                           # découpe en liste de mots (par défaut sur les espaces)
````

##### **4. Listes et dictionnaires**

- Une **liste** `[]` est une collection ordonnée et modifiable.
- Un **dictionnaire** `{}` associe des clés à des valeurs — parfait pour compter des mots !

````python
mots = phrase.lower().strip().split()
print(mots)
print(len(mots))       # nombre d'éléments
print(mots[0])         # premier élément
print(mots[-1])        # dernier élément

frequences = {}
for mot in mots:
    if mot in frequences:
        frequences[mot] += 1
    else:
        frequences[mot] = 1

print(frequences)
````

**La méthode `.count()`**

La méthode `.count()` s'applique sur une liste d'éléments et permet de compter le nombre d'éléments égaux à une chaîne cible.

````python
nb_automatique = phrase.split().count("Automatique")
print(nb_automatique)
````

**La fonction `max()`**

La fonction `max()`, associé à un attribut `key` peut s'appliquer sur une liste d'éléments pour trouver l'élément qui a l'attribut `key` le plus grand. Cela peut par exemple servir à trouver le mot le plus long dans une phrase.

````python
mot_plus_long = max(mots, key=len)
print(mot_plus_long)
````

##### **5. Conditions et boucles**

Pour itérer sur les éléments d'une liste, on peut utiliser une boucle "`for` ... `in` ..." et éventuellement ajouter des conditions à appliquer sur chacun de ses éléments avec `if`/`else`.

````python
# Exemple : trier les mots en fonction de leur longueur (len)
for mot in mots:
    if len(mot) > 5:
        print(f"{mot} -> mot long")
    else:
        print(f"{mot} -> mot court")
````

**Listes en comprehension (très utilisées en NLP)**

Façon compacte d'écrire une boucle qui construit une liste.

````python
mots_longs = [mot for mot in mots if len(mot) > 5]
print(mots_longs)
````

##### **6. Fonctions**

On encapsule un traitement réutilisable dans une fonction qu'on définit avec `def`.

````python
def nettoyer_texte(texte):
    texte = texte.lower().strip()
    mots = texte.split()
    return mots

resultat = nettoyer_texte("  Encore un Exemple DE Phrase à Nettoyer !  ")
print(resultat)
````

**Traitement du texte avec spaCy**

La première étape lorsque l'on souhaite traiter du texte automatiquement avec spaCy, est de charger le modèle de la langue de notre corpus et de l'appliquer au corpus.

````python
# Charger le modèle français (celui qu'on a téléchargé juste avant)
nlp = spacy.load('fr_core_news_lg')

# Traiter le texte avec spaCy
texte = "Ceci est le texte que je vais traiter avec spaCy."
doc = nlp(texte)
````

Lors du traitement (quand on fait `doc = nlp(texte)`), spaCy commence par **tokeniser** le texte et attribue à chacun de ces tokens des propriétés que nous verrons au fur et à mesure.

Attention ! Le traitement est effectué en appliquant des **règles spécifiques à chaque langue**. Il est donc important de bien sélectionner le modèle de traitement de la langue de notre corpus ! Dans notre cas, on a choisi le modèle `fr_core_news_lg` puisque notre corpus est en français.

Les données d'entraînement utilisées pour l'entraînement du modèle sont indiquées [ici](https://spacy.io/models/fr).

Après application du modèle à notre texte, les tokens sont stockés dans dans l'objet `doc`.

**Named Entity Recognition (_NER_)**

Pour extraire les entités nommées d'un texte, on utilise l'attribut `.ents`.

````python
texte = "Le juge Wargrave, retraité depuis peu, tirait sur son cigare en parcourant avec intérêt les pages politiques du Times."
doc = nlp(texte)
named_entities = [(ent.text, ent.label_) for ent in doc.ents]
print(named_entities)
````

**Remarque 2** : SpaCy permet d'ajouter sa propre liste d'entités nommées, avec le module `EntityRuler`. Une entité nommée, de manière générale, désigne tout élément qui peut être référencé par un nom propre, comme une personne, un lieu ou une organisation. Mais on peut tout à fait étendre ce concept dans le cadre d'un projet de recherche !

> **Exemple** : On souhaite étudier la présence des drogues dans les romans policiers. On peut alors établir une liste des mots correspondant aux concepts que l'on souhaite relever, et relever ces derniers, plutôt que les personnes, lieux, etc.

````python
from spacy.pipeline import EntityRuler

# Fonction d'ajout d'entités nommées
def add_entities(dico):
    nlp = spacy.load("fr_core_news_lg")
    ruler = nlp.add_pipe("entity_ruler", after="ner", config={"overwrite_ents": True})
    patterns = [
        {"label": label, "pattern": mot}
        for mot, label in dico.items()
    ]
    ruler.add_patterns(patterns)
    return nlp

# Dictionnaire des entités à ajouter
dico_entites = {
    "cigarette": "DROGUE",
    "cigare": "DROGUE",
    "pipe": "DROGUE",
    "alcool": "DROGUE",
    "vin": "DROGUE"
}

# Ajout
nlp_drugs = add_entities(dico_entites)

# Application du nouveau modèle à la phrase
texte_annote_avec_drogues = nlp_drugs(texte)
displacy.render(texte_annote_avec_drogues, style="ent", jupyter=True)
````

**Syntaxe en dépendance**

Pour extraire la structure syntaxique d'une phrase, on peut utiliser l'attribut `.dep` qui indique la relation de dépendance de chaque token, et l'attribut `.head` qui correspond à son gouverneur.

````python
texte = "Un groupe d'amis fait la fête sur la plage."
doc = nlp(texte)
for token in doc:
    print(f"{token.text} → {token.dep_} ({token.head})")
````

Encore une fois, ce n'est pas hyper clair, mais le module `displacy` propose une représentation bien plus visuelle, sous forme d'arbre.

````python
displacy.render(doc, style='dep')
````

On peut également utiliser divers attributs pour parcourir cet arbre de dépendances :

- `subtree` : tous les descendants (directs ou non) d'un token
- `children` : tous les enfants (descendants directs) d'un token
- `rights` : tous les enfants à droite d'un token
- `lefts` : tous les enfants à gauche d'un token
- `ancestors` : tous les ancêtres d'un token

````python
# Exemple avec "fête" comme token
ref_token = [token for token in doc if token.text == "fête"][0]

# Descendants
print(f"Descendants de \"{ref_token}\" : {list(ref_token.subtree)}")

# Enfant
print(f"Enfants de \"{ref_token}\" : {list(ref_token.children)}")

# Enfants à droite
print(f"Enfants à droite de \"{ref_token}\" : {list(ref_token.rights)}")

# Enfants à gauche
print(f"Enfants à gauche de \"{ref_token}\" : {list(ref_token.lefts)}")

# Ancêtres
print(f"Parents de \"{ref_token}\" : {list(ref_token.ancestors)}")
````

On peut par exemple utiliser ces propriétés pour reconstituer la structure globale de la phrase (sujet, verbe, objet).

````python
# Extraction de la racine
root = [token for token in doc if token.head == token][0]

# Extraction du sujet
sujet = [token for token in root.children if token.dep_ == "nsubj"][0]

# Extraction de l'objet
objet = [token for token in root.children if token.dep_ == "obj"][0]

# Affichage
print(f"{sujet} (S), {root} (V), {objet}(O)")
````

Mais attention ! Il ne faut jamais perdre de vue qu'un même personnage peut posséder des dénominations différentes et être alors compté comme plusieurs personnages.
